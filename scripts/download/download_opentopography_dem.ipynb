# Copernicus DEM (OpenTopography)

Needs a free API key: https://portal.opentopography.org/myopentopo
Set `OPENTOPOGRAPHY_API_KEY` in `.env` (see `.env.example`).

```python
# make sure we're running from scripts/download so `import common` resolves
import os, sys

NOTEBOOK_DIR = os.path.abspath(".")
if os.path.basename(NOTEBOOK_DIR) != "download":
    NOTEBOOK_DIR = os.path.join(NOTEBOOK_DIR, "scripts", "download")
os.chdir(NOTEBOOK_DIR)
if NOTEBOOK_DIR not in sys.path:
    sys.path.insert(0, NOTEBOOK_DIR)
print("cwd:", os.getcwd())
```

```python
import os

from common import BANTEN_BBOX, RAW_DIR, download_file, ensure_dir, get_logger

log = get_logger("opentopography")

OUT_DIR = RAW_DIR / "dem"
API_URL = "https://portal.opentopography.org/API/globaldem"
DEM_TYPE = "COP30"  # Copernicus GLO-30


def main() -> None:
    api_key = os.getenv("OPENTOPOGRAPHY_API_KEY")
    if not api_key:
        log.error(
            "OPENTOPOGRAPHY_API_KEY isn't set. Sign up for free at "
            "https://portal.opentopography.org/myopentopo and add it to .env"
        )
        return

    ensure_dir(OUT_DIR)
    params = {
        "demtype": DEM_TYPE,
        "south": BANTEN_BBOX["south"],
        "north": BANTEN_BBOX["north"],
        "west": BANTEN_BBOX["west"],
        "east": BANTEN_BBOX["east"],
        "outputFormat": "GTiff",
        "API_Key": api_key,
    }
    query = "&".join(f"{k}={v}" for k, v in params.items())
    url = f"{API_URL}?{query}"

    dest = OUT_DIR / f"{DEM_TYPE.lower()}_banten.tif"
    download_file(url, dest)
    log.info("Done. DEM saved to %s", dest)
```

```python
main()
```
