# WorldClim

BIO1 (temperature), BIO12 (precipitation), BIO15 (seasonality) and the rest - 2.5 arc-minute resolution, long-term climatology (not tied to a specific year).

```python
# make sure we're running from scripts/download so `import common` resolves
import os, sys

NOTEBOOK_DIR = os.path.abspath(".")
if os.path.basename(NOTEBOOK_DIR) != "download":
    NOTEBOOK_DIR = os.path.join(NOTEBOOK_DIR, "scripts", "download")
os.chdir(NOTEBOOK_DIR)
if NOTEBOOK_DIR not in sys.path:
    sys.path.insert(0, NOTEBOOK_DIR)
print("cwd:", os.getcwd())
```

```python
import zipfile

from common import RAW_DIR, download_file, ensure_dir, get_logger

log = get_logger("worldclim")

RESOLUTION = "2.5m"  # other options: 30s, 5m, 10m (finer = bigger files)
URL = f"https://geodata.ucdavis.edu/climate/worldclim/2_1/base/wc2.1_{RESOLUTION}_bio.zip"

OUT_DIR = RAW_DIR / "worldclim"


def main() -> None:
    ensure_dir(OUT_DIR)
    zip_path = OUT_DIR / f"wc2.1_{RESOLUTION}_bio.zip"
    download_file(URL, zip_path)

    log.info("Extracting %s", zip_path)
    with zipfile.ZipFile(zip_path) as zf:
        zf.extractall(OUT_DIR)
    log.info("Done. GeoTIFFs are in %s (look for *_bio_1.tif, *_bio_12.tif, *_bio_15.tif)", OUT_DIR)
```

```python
main()
```
